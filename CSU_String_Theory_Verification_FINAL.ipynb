```python
# ═══════════════════════════════════════════════════════════════════════════════
# SECTION 13: VIRASORO OPERATORS
# ═══════════════════════════════════════════════════════════════════════════════
print("\n" + "═" * 80)
print("SECTION 13: VIRASORO ALGEBRA WITH sympy.physics.quantum")
print("═" * 80)

# Define Virasoro generators as Operators
print("\n► Creating Virasoro operators L_n:")

# Create operator class for Virasoro
class VirasoroOp(Operator):
    """Virasoro generator L_n"""
    def __new__(cls, n):
        return Operator.__new__(cls, n)
    
    @property
    def n(self):
        return self.args[0]

# Create some Virasoro operators
m_sym = Symbol('m', integer=True)
n_sym = Symbol('n', integer=True)

L_m = VirasoroOp(m_sym)
L_n = VirasoroOp(n_sym)
L_0 = VirasoroOp(0)
L_1 = VirasoroOp(1)
L_minus1 = VirasoroOp(-1)
L_2 = VirasoroOp(2)
L_minus2 = VirasoroOp(-2)

print(f"  L_m = {L_m}")
print(f"  L_n = {L_n}")
print(f"  L_0 = {L_0}")
print(f"  L_1 = {L_1}")
print(f"  L_{-1} = {L_minus1}")

# Virasoro commutator (formal)
print("\n► Virasoro commutation relation:")
comm_Lm_Ln = Commutator(L_m, L_n)
print(f"  [{L_m}, {L_n}] = {comm_Lm_Ln}")

# Full algebra
c_sym = Symbol('c')  # Central charge
print("\n► Full Virasoro algebra:")
print("  [L_m, L_n] = (m - n)L_{m+n} + (c/12)m(m² - 1)δ_{m+n,0}")

verify(True, "Virasoro operators defined using sympy.physics.quantum.Operator")
```

```python
# ═══════════════════════════════════════════════════════════════════════════════
# SECTION 14: VIRASORO CENTRAL TERM VERIFICATION
# ═══════════════════════════════════════════════════════════════════════════════
print("\n" + "═" * 80)
print("SECTION 14: VIRASORO CENTRAL TERM c/12 · m(m²-1)")
print("═" * 80)

# The central term in Virasoro algebra
c_sym = Symbol('c')  # Central charge
m_sym2 = Symbol('m', integer=True)

# Central extension term
central_term = (c_sym / 12) * m_sym2 * (m_sym2**2 - 1)
central_expanded = expand(central_term)
print(f"\n► Central term: (c/12)·m·(m²-1)")
print(f"  Expanded: {central_expanded}")

# Factored form
central_factored = (c_sym / 12) * m_sym2 * (m_sym2 - 1) * (m_sym2 + 1)
print(f"  Factored: (c/12)·m·(m-1)·(m+1)")

# Verify for specific m values
print("\n► Central term for specific m values:")
for m_val in [-2, -1, 0, 1, 2, 3]:
    term = central_term.subs(m_sym2, m_val)
    term_simplified = simplify(term)
    print(f"  m = {m_val:2d}: (c/12)·{m_val}·({m_val}²-1) = {term_simplified}")

# Key observation: central term vanishes for m = -1, 0, 1
print("\n► Important: Central term vanishes for m ∈ {-1, 0, 1}")
print("  These generate the global conformal group SL(2,R)")

for m_val in [-1, 0, 1]:
    term = central_term.subs(m_sym2, m_val)
    verify(simplify(term) == 0, f"Central term vanishes for m = {m_val}")
```
